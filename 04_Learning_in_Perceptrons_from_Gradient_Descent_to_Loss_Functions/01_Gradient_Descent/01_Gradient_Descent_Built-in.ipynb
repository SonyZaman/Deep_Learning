```python
from sklearn.datasets import make_regression
import matplotlib.pyplot as plt
import numpy as np
from sklearn.model_selection import cross_val_score
```

**`make_regression(...)`**: This is the function being called to generate the dataset. Let's look at its arguments:

*   **`n_samples=100`**: This tells the function to generate **100 data points** (or rows) in your dataset.

*   **`n_features=1`**: This specifies that there will be only **1 feature** (or column) in `X`. This is useful for simple linear regression (e.g., predicting weight based solely on height).

*   **`n_informative=1`**: This means that **1 feature** is actually useful (informative) for predicting the target `y`. Since `n_features` is also 1, it means the single feature we generated is directly related to the output. (If `n_features` was 5 and `n_informative` was 1, it would generate 4 columns of useless random noise).
*   **`n_targets=1`**: This specifies that there is only **1 target variable** (`y`) to predict.

*   **`noise=20`**: This adds Gaussian (normal) noise to the output `y`. A value of `20` means the data points will be somewhat scattered around the true linear relationship, making it look more like real-world, imperfect data rather than a perfectly straight line.

*   **`random_state=13`**: This is a seed for the random number generator. By setting it to a specific number (like `13`), you ensure that **every time you run this exact line of code, you will get the exact same random dataset**. This is very useful for reproducibility when learning or testing models.


```python
X,y = make_regression(n_samples=100, n_features=1, n_informative=1, n_targets=1,noise=20,random_state=13)
```

*   **`print(X.shape)`**: Prints the dimensions (number of rows and columns) of the input features `X`. In this case, it will output `(100, 1)`, meaning 100 samples and 1 feature.

```python
print(X.shape)
```

*   **`print(y.shape)`**: Prints the dimensions of the target variable `y`. It will output `(100,)`, indicating it's a 1-dimensional array with 100 values corresponding to the 100 samples.

```python
print(y.shape)
```

*   **`plt.scatter(X, y)`**: Creates a scatter plot visualizing the relationship between the features `X` (on the x-axis) and the target `y` (on the y-axis), displaying each data point as a dot.

```python
plt.scatter(X,y)
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

**`lr = LinearRegression()`**: This creates an instance (or object) of the Linear Regression model and stores it in the variable `lr`. At this point, the model is "empty" or untrained; it hasn't seen any data yet.

**`lr.fit(X_train, y_train)`**: This is the crucial training step. The `fit` method tells the model to learn the relationship between the training inputs (`X_train`) and the training outputs (`y_train`). For linear regression, it calculates the optimal line (the slope and y-intercept) that best fits this data.

**`print(lr.coef_)`**: After the model is trained, `lr.coef_` stores the learned **coefficient** (often represented as $m$ or $w$). In a simple linear regression formula ($y = mx + c$), this is the slope of the line. It tells you how much the predicted `y` increases for every 1-unit increase in `X`. 

**`print(f"{lr.coef_}X + {lr.intercept_}")`**: This prints the final mathematical equation that the model learned.
*   **`lr.coef_`**: As mentioned, this is the slope ($m$).
*   **`lr.intercept_`**: This stores the learned **intercept** ($c$ or $b$). It represents the value of $y$ when $X$ is exactly 0. 

So, this line prints the learned formula in the format: **$y = mX + c$**, showing exactly how the model will make future predictions.


```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(X_train,y_train)


print(lr.coef_)
print(f"{lr.coef_}X + {lr.intercept_}")
```

**`y_pred = lr.predict(X_test)`**: Now that the model (`lr`) has been trained, we use the `predict` method to test it. We feed it the test features (`X_test`), which it hasn't seen during training, and ask it to predict the corresponding target values. These predicted values are saved in the variable `y_pred`.

**`r2_score(y_test, y_pred)`**: This calculates the R-squared ($R^2$) score to evaluate how well the model performed. It compares the true, actual values from our test set (`y_test`) against the values our model predicted (`y_pred`). 
*   An $R^2$ score of `1.0` is perfect (the model's predictions perfectly match the actual data).
*   A score closer to `0` means the model fails to capture the trend of the data. 

```python
from sklearn.metrics import accuracy_score
from sklearn.metrics import r2_score

y_pred = lr.predict(X_test)

r2_score(y_test,y_pred)
```

**`print(y_pred)`**: This simply prints out the array of numbers that our model predicted, allowing you to manually inspect the output values.
