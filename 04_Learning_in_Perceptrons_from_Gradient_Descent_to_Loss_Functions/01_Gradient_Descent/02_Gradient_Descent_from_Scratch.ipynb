
This code defines a custom class `GDRegressor` that builds a Linear Regression model from scratch using the **Gradient Descent** optimization algorithm, instead of relying on scikit-learn's built-in functions.


```python
class GDRegressor:
  def __init__(self,learning_rate,epochs):

    self.m = 1
    self.b = 0
    self.lr = learning_rate
    self.epochs = epochs

  def fit(self,X,y):
    #calculate b

    for i in range(self.epochs ):
      loss_slope_b = -2*np.sum( y - self.m*X.ravel() - self.b )
      loss_slope_m = -2 * np.sum((y - self.m*X.ravel() - self.b)*X.ravel())

      # update parameters
      self.b = self.b - (self.lr*loss_slope_b)
      self.m = self.m - (self.lr * loss_slope_m)



      print(f" m = {self.m}, b = {self.b}")

  def predict(self,X):
    return self.m * X +self.b
```

```python
gd = GDRegressor(0.001,50)
gd.fit(X_train,y_train)
```

```python
y_pred_custom = gd.predict(X_test)
print(y_pred_custom)
```
